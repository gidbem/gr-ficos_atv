```python
import seaborn as sns
import matplotlib.pyplot as plt

# Carregando a base do Titanic
df_titanic = sns.load_dataset('titanic')
display(df_titanic .head())
```

```python
#QUESTÃO 1

plt.figure(figsize=(8, 4))
sns.histplot(data=df_titanic, x='fare', kde=True, bins=40, color='purple')

plt.title('Distribuição de Tarifas')
plt.xlabel('Tarifa')
plt.ylabel('Frequência')
plt.show()

#a maioria pagou tarifas baratas
#a distribuição é assimétrica
```

```python
#QUESTÃO 2

plt.figure(figsize=(8, 4))
sns.boxplot(data=df_titanic, x='sex', y='age')

plt.title('Comaprativo de Idade por Gênero')
plt.xlabel('Gênero')
plt.ylabel('Idade')
plt.show()

#o gênero masculino apresentava maior mediana
```

```python
#QUESTÃO 3

plt.figure(figsize=(8, 4))
sns.regplot(data=df_titanic, x='age', y='fare')

plt.title('Tendência visual')
plt.xlabel('Idade')
plt.ylabel('Tarifa')
plt.show()

#a linha é ascendente
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Carregando a base de carros
df_mpg = sns.load_dataset('mpg')
display(df_mpg)
```

```python
#QUESTÃO 1.2

plt.figure(figsize=(8, 4))
sns.scatterplot(data=df_mpg, x='horsepower', y='mpg', hue='origin', palette={'usa':'pink', 'japan':'red', 'europe':'blue'},alpha=0.7)

plt.title('Gráfico de Dispersão (Potência vs Consumo)')
plt.xlabel('Potência')
plt.ylabel('Consumo')
plt.show()

#Não. Carros mais potentes tendem a consumir menos gasolina.
```

```python
#QUESTÃO 2.2 

plt.figure(figsize=(8,4))
sns.lineplot(data=df_mpg, x='model_year', y='mpg')

plt.title('Evolução de eficiência')
plt.xlabel('Modelos')
plt.ylabel('Consumo médio')
plt.show()

#Sim, a eficiência melhorou ao longo dos anos. 
```

```python
#QUESTÃO 3.2

plt.figure(figsize=(8,4))
sns.boxplot(data=df_mpg, x='origin', y='acceleration')

plt.title('Distribuição do tempo de aceleração por Região')
plt.xlabel('Origem do veículo')
plt.ylabel('Aceleração')
plt.show()
```

```python
#Verificando os tipos de dados
df_mpg.dtypes
```

```python
#Seperando os dados numericos
df_mum_mpg1 = df_mpg[['mpg','cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year']]

df_mum_mpg2 = df_mpg.select_dtypes(['float', 'int64'])

display(df_mum_mpg1)
display(df_mum_mpg2)
```

```python
#calculando a matriz de correlação
matriz_corr = df_mum_mpg1.corr()

#Criando o mapa de calor
sns.heatmap(matriz_corr, annot=True, fmt='.2f', cmap='coolwarm')

#grafico
plt.title('Matriz de correlação')

plt.show()
```
