```python
import seaborn as sns
import matplotlib.pyplot as plt

# Carregando a base do Titanic
df_titanic = sns.load_dataset('titanic')
display(df_titanic .head())
```

```python
#QUESTÃO 1

plt.figure(figsize=(8, 4))
sns.histplot(data=df_titanic, x='fare', kde=True, bins=40, color='purple')

plt.title('Distribuição de Tarifas')
plt.xlabel('Tarifa')
plt.ylabel('Frequência')
plt.show()

#a maioria pagou tarifas baratas
#a distribuição é assimétrica
```

```python
#QUESTÃO 2

plt.figure(figsize=(8, 4))
sns.boxplot(data=df_titanic, x='sex', y='age')

plt.title('Comaprativo de Idade por Gênero')
plt.xlabel('Gênero')
plt.ylabel('Idade')
plt.show()

#o gênero masculino apresentava maior mediana
```

```python
#QUESTÃO 3

plt.figure(figsize=(8, 4))
sns.regplot(data=df_titanic, x='age', y='fare')

plt.title('Tendência visual')
plt.xlabel('Idade')
plt.ylabel('Tarifa')
plt.show()

#a linha é ascendente
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Carregando a base de carros
df_mpg = sns.load_dataset('mpg')
display(df_mpg.head())
```

```python
#QUESTÃO 1.2

plt.figure(figsize=(8, 4))
sns.scatterplot(data=df_mpg, x='horsepower', y='mpg', hue='origin', palette={'usa':'pink', 'japan':'red', 'europe':'blue'},alpha=0.7)

plt.title('Gráfico de Dispersão (Potência vs Consumo)')
plt.xlabel('Potência')
plt.ylabel('Consumo')
plt.show()

#Não. Carros mais potentes tendem a consumir menos gasolina.
```

```python
#QUESTÃO 2.2 

plt.figure(figsize=(8,4))
sns.lineplot(data=df_mpg, x='model_year', y='mpg')

plt.title('Evolução temporal')
plt.xlabel('Modelos')
plt.ylabel('Evolução média')
plt.show()

#Sim, a eficiência melhorou ao longo dos anos. 
```

```python
#QUESTÃO 3.2

plt.figure(figsize=(8,4))
sns.boxplot(data=df_mpg, x='acceleration', y='origin')

plt.title('Detecção de Outliners')
plt.xlabel('Acelereção')
plt.ylabel('País de origem')
plt.show()
```
